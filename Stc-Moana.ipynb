```python
import pandas as pd
import warnings

warnings.filterwarnings('ignore')

file_path = "stc TV Data Set_T3.xlsx"
df = pd.read_excel(file_path)

display(df.head())
```

```python
user_item_matrix = df.groupby(['user_id_maped', 'program_name'])['rating'].max().unstack()
```

```python
moana_ratings = user_item_matrix['Moana']
similar_to_moana = user_item_matrix.corrwith(moana_ratings)

corr_moana = pd.DataFrame(similar_to_moana, columns=['Correlation']).dropna()

ratings_count = df.groupby('program_name')['rating'].count()
corr_moana = corr_moana.join(ratings_count.rename('ratings_count'))
```

Top 5 recommendations for users who watched the movie "Moana"

```python
top_5_recommendations = corr_moana[corr_moana['ratings_count'] > 500].sort_values(by='Correlation', ascending=False)

display(top_5_recommendations.iloc[1:6])
```
